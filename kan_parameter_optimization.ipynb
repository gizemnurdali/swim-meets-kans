```python
import utils
import importlib
import evaluation_metrics as em
from datasets import load_dataset
importlib.reload(utils)
```

# TF1

```python
# Load TF1 dataset
tf1 = load_dataset("TF1")
tf1_x_train, tf1_x_test, tf1_y_train, tf1_y_test = tf1["train_input"], tf1["test_input"], tf1["train_label"], tf1["test_label"]
```

```python
tf1_kan_study = utils.study_optuna_kan("TF1", tf1_x_train, tf1_y_train, tf1_x_test, tf1_y_test)
```

```python
tf1_best_arch = tf1_kan_study["architecture"]
tf1_best_arch
```

```python
tf1_kan_final_model = utils.build_kan(width=tf1_best_arch, grid=3, k=3, seed=42)
tf1_kan_final_model = utils.fit_kan_with_grid_extension(
    tf1_kan_final_model,
    tf1_x_train, tf1_y_train, tf1_x_test, tf1_y_test,
    grids=(5, 10, 20), steps_per_grid=20, opt="LBFGS", lamb=0,
)

tf1_kan_results = em.evaluate(
    utils.KANModel(tf1_kan_final_model),
    tf1_x_train, tf1_y_train, tf1_x_test, tf1_y_test
)
em.print_results(tf1_kan_results)
```
